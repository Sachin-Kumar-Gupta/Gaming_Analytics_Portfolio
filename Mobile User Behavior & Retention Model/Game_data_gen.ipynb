```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)
n_players = 10000

data = {
    'player_id': range(1, n_players + 1),
    'level_reached': np.random.randint(1, 50, n_players),
    'session_duration': np.random.normal(15, 5, n_players).clip(1, 60), # Avg 15 mins
    'in_app_purchase_amount': np.random.exponential(10, n_players).round(2)
}

df = pd.DataFrame(data)

# Logic: Players with lower levels and shorter sessions are more likely to churn
# Churn probability calculation
logit = (20 - df['level_reached'] - df['session_duration']) / 10
prob = 1 / (1 + np.exp(-logit))
df['churned'] = np.random.binomial(1, prob)

# Save to CSV
df.to_csv('gaming_churn_data.csv', index=False)
print("Dataset created: gaming_churn_data.csv")
```
